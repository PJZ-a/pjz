```python
import os
import zipfile

root = r'C:\Users\86199\Desktop\个人主页'
output = os.path.join(root, 'pan-jianze-homepage.zip')
files = ['index.html', 'styles.css', 'README.md', '简历.docx', '.nojekyll']
assets_dir = os.path.join(root, 'assets')
with zipfile.ZipFile(output, 'w', zipfile.ZIP_DEFLATED) as z:
    for fn in files:
        path = os.path.join(root, fn)
        if os.path.exists(path):
            z.write(path, fn)
        else:
            print('Missing', fn)
    if os.path.isdir(assets_dir):
        for dirpath, _, filenames in os.walk(assets_dir):
            for fname in filenames:
                full = os.path.join(dirpath, fname)
                z.write(full, os.path.relpath(full, root))
    else:
        print('Missing assets directory')
print('Created:', output)
```
